```python
import json
import glob
```

```python
CONVO_DIR = "/Users/siyanli/Documents/CITRUS/game_user_study_pre_filter/recordings-download_batch_1"

for move_file in glob.glob(CONVO_DIR + "/*/moves.jsonl"):
    pid = move_file.split("/")[-2]
    print(pid)
    all_moves = [json.loads(l) for l in open(move_file).readlines()]
    if len(all_moves) > 8:
        print("EXCLUDED")
        print("----")
        continue
    all_optim = sum([1 for m in all_moves if m["optimal"]])
    move_eff = all_optim / len(all_moves)
    print("SOLVED:", all_optim == 5)
    print("MOVE EFFICIENCY:", move_eff)
    print("----")
```
