```python
import pandas

baseline_adopters = pandas.read_csv("scores_adopters.csv")
baseline_na = pandas.read_csv("scores.csv")
```

# Filter Participants

```python
def filter_participants(df: pandas.DataFrame):
    fin_list = []
    for c in df.pid.unique():
        curr_df = df.loc[df["pid"] == c]
        if len(curr_df) < 3:
            continue
        fin_list.append(c)
    return fin_list

fin_adopters = filter_participants(baseline_adopters)
fin_non_adopters = filter_participants(baseline_na)
```

```python
len(fin_adopters), len(fin_non_adopters)
```

```python
baseline_adopters = baseline_adopters.loc[baseline_adopters.pid.isin(fin_adopters)]
baseline_na = baseline_na.loc[baseline_na.pid.isin(fin_non_adopters)]
```

```python
from scipy.stats import mannwhitneyu
import numpy as np

def compare_rounds(rr: str):
    rx_adopters = baseline_adopters.loc[baseline_adopters["round"] == rr]["margin"].dropna().to_list()
    rx_non_adopters = baseline_na.loc[baseline_na["round"] == rr]["margin"].dropna().to_list()
    print("COMPARING: ", rr)
    print("\t ADOPTER MEAN:", np.mean(rx_adopters))
    print("\t NON-ADOPTER MEAN:", np.mean(rx_non_adopters))
    print(mannwhitneyu(rx_adopters, rx_non_adopters))

for r in ["R1", "R2", "R3"]:
    compare_rounds(r)
```
